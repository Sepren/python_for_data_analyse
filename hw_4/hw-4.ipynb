```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# 1.Считайте датасет из файла train.csv (это данные о выживаемости на Титанике)

```python
df = pd.read_csv('/kaggle/input/titanic/train.csv')
df
```

# 2.Визуализируйте распределение значений признаков Survived, Pclass, Age, Sex. Parch

```python
small_df = df[['Survived', 'Pclass', 'Age', 'Sex', 'Parch']]
small_df
```

1. Рассмотрим график зависимости Survived и Age ввиде ящика с усами

```python
small_df_sa = small_df[['Survived', 'Age']]
```

```python
# sns.boxplot(small_df_sa, hue="Survived")
# small_df_sa = sns.load_dataset(small_df_sa)
sns.boxplot(
    x="Survived",
    y="Age",
    data=small_df_sa
)
plt.title('Age Distribution by Survival Status')
plt.legend(title='Survived', labels=['Did Not Survive', 'Survived'])
plt.show()
```

```python
# histplot (попытка реализовать через гистограмму, ну такое)
sns.histplot(x="Survived", y="Age", data=small_df_sa)
plt.title('Age Distribution by Survival Status')
plt.legend(title='Survived', labels=['Did Not Survive', 'Survived'])
plt.show()
```

2. Рассмотрим график зависимости PClass и Age ввиде ящика с усами

```python
small_df_pa = small_df[['Pclass', 'Age']]
small_df_pa
```

```python
sns.boxplot(x="Pclass", y="Age", data=small_df_pa)
plt.title('Age Distribution by Pclass Status')
plt.legend(title='Pclass', labels=['1', '2', '3'])
plt.show()
```

3. Рассмотрим график зависимости Sex и Age ввиде ящика с усами

```python
small_df_sexa = small_df[['Sex', 'Age']]
small_df_sexa
```

```python
sns.boxplot(x="Sex", y="Age", data=small_df_sexa)
plt.title('Age Distribution by Sex')
plt.legend(title='Pclass', labels=['male', 'female'])
plt.show()
```

4. Рассмотрим график зависимости Parch и Age ввиде ящика с усами

```python
small_df_parcha = small_df[['Parch', 'Age']]
small_df_parcha
```

```python
sns.boxplot(x="Parch", y="Age", data=small_df_parcha)
plt.title('Age Distribution by Parch')
plt.legend(title='Parch', labels=['0', '1', '2', '3', '4', '5', '6'])
plt.show()
```

```python
# Попытка реализовать ввиде гистограммы
sns.histplot(x="Parch", y="Age", data=small_df_parcha)
plt.title('Age Distribution by Parch')
plt.legend(title='Parch', labels=['0', '1', '2', '3', '4', '5', '6'])
plt.show()
```

# 5.Постройте график типа pie chart для переменных Survived, Pclass, подпишите доли в процентах

```python
small_df_sp = df[['Survived', 'Pclass']]
small_df_sp
```

```python
group_df = small_df_sp.groupby('Survived').count()
group_df
```

```python
group_df.index
```

```python
plt.pie(group_df['Pclass'], labels = group_df.index)
plt.xlabel("Значение по оси X") 
plt.ylabel("Значение по оси Y") 
plt.title('Survived Distribution by Pclass')
plt.legend(title='Pclass', labels=['0', '1'])
plt.show()
```

# 7.Постройте интерактивный sunburst plot (визуализация иерархических данных) с помощью plotly.

```python
import plotly.express as px

fig = px.sunburst(df, path=['Pclass', 
                            'Survived',
                            'Parch'], 
                  values='Age')
fig.update_layout(
    showlegend=True,
    legend_title_text='Statistics',
    title_x=0.5
)
fig.show()
```
