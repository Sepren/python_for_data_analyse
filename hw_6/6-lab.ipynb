```python
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.metrics import roc_auc_score
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, f1_score
```

```python
import polars as pl
```

```python
df = pl.read_csv('/kaggle/input/titanic-dataset/Titanic-Dataset.csv')
df
```

```python
new_df = df.clone()
new_df = new_df.drop(['Name'])
```

```python
new_df = new_df.drop(['Ticket', 'Cabin'])
new_df
```

```python
new_df = new_df.to_dummies("Sex")
```

```python
new_df = new_df.to_dummies("Embarked")
```

```python
X = new_df.drop(['Survived'])
X
```

```python
y = new_df['Survived']
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42, stratify=y)
X_train
```

```python
model = XGBClassifier()
```

```python
# log_reg = LogisticRegression()
model.fit(X_train, y_train)
```

```python
y_pred_xgboost = model.predict(X_test)
accuracy_score(y_test, y_pred_xgboost)
```

```python
f1_score(y_test, y_pred_xgboost)
```

Был произведен сравнительный анализ между f1_score и accuracy_score. Так как у нас задача с классификацией в виде предсказания класса, поэтому рассматривались такие метрики. Accuracy_score показал более лучшую точность предсказания, чем f1_score
