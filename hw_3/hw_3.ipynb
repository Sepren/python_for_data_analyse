Выведите самое популярное мужское и самое популярное женское имя на корабле в каждом классе - 2 балла


```python
def parse_first_name_female(name):
    first = name.str.extract(r"Mrs\.\s+[^(]*\((\w+)", expand=False)
    first.loc[first.isna()] = name.str.extract(r"\.\s+(\w+)", expand=False)
    return first
```

```python
female_name = parse_first_name_female(df.loc[df["Sex"]=="female", "Name"])
female_name.value_counts()
```

```python
def parse_first_name_male(name):
    first = name.str.extract(r"Mr\.\s+[^(]*\((\w+)", expand=False)
    first.loc[first.isna()] = name.str.extract(r"\.\s+(\w+)", expand=False)
    return first
```

```python
male_name = parse_first_name_male(df.loc[df["Sex"]=="male", "Name"])
male_name.value_counts()
```
